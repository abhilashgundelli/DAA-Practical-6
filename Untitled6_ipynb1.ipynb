### Chain Matrix Multiplication using Dynamic Programming

This problem aims to find the most efficient way to multiply a given sequence of matrices. The problem is not actually to perform the multiplications, but merely to decide the order in which to perform them. The most common solution uses dynamic programming to minimize the number of scalar multiplications.

```python
def matrix_chain_order(p):
    """
    Calculates the minimum number of scalar multiplications needed
    to multiply a chain of matrices.

    Args:
        p (list): A list of dimensions for the matrices.
                  p[i-1] x p[i] is the dimension of matrix i.

    Returns:
        int: The minimum number of scalar multiplications.
        list: The optimal parenthesization of the matrix chain.
    """
    n = len(p) - 1  # Number of matrices

    # m[i][j] stores the minimum number of scalar multiplications
    # needed to compute the matrix A[i]...A[j]
    m = [[0 for _ in range(n)] for _ in range(n)]

    # s[i][j] stores the index k such that A[i]...A[j] is split at k
    s = [[0 for _ in range(n)] for _ in range(n)]

    # l is chain length
    for l in range(2, n + 1):
        for i in range(n - l + 1):
            j = i + l - 1
            m[i][j] = float('inf')
            for k in range(i, j):
                # q = cost of multiplying matrices A[i]...A[k] + A[k+1]...A[j]
                # + cost of multiplying the two resulting matrices
                q = m[i][k] + m[k + 1][j] + p[i] * p[k + 1] * p[j + 1]
                if q < m[i][j]:
                    m[i][j] = q
                    s[i][j] = k

    return m[0][n - 1], s

def print_optimal_parenthesization(s, i, j):
    """
    Prints the optimal parenthesization of the matrix chain.

    Args:
        s (list): The s table from matrix_chain_order.
        i (int): Starting index of the current subchain.
        j (int): Ending index of the current subchain.
    """
    if i == j:
        return f"A{i+1}"
    else:
        k = s[i][j]
        left = print_optimal_parenthesization(s, i, k)
        right = print_optimal_parenthesization(s, k + 1, j)
        return f"({left}{right})"


# Example Usage:
# p represents the dimensions of the matrices.
# For example, p = [10, 20, 30] means two matrices:
# A1: 10x20 and A2: 20x30
# For p = [10, 20, 30, 40] means three matrices:
# A1: 10x20, A2: 20x30, A3: 30x40

dimensions = [30, 35, 15, 5, 10, 20, 25]
min_multiplications, split_points = matrix_chain_order(dimensions)

print(f"Minimum number of scalar multiplications: {min_multiplications}")

optimal_parenthesization = print_optimal_parenthesization(split_points, 0, len(dimensions) - 2)
print(f"Optimal parenthesization: {optimal_parenthesization}")
```

### Explanation:

1.  **`matrix_chain_order(p)` function:**
    *   `n` is the number of matrices. If `p` has `n+1` elements, there are `n` matrices. For example, `[A, B, C, D]` means 4 matrices, thus `n=4`. `p = [p0, p1, p2, p3, p4]`. A0 is `p0 x p1`, A1 is `p1 x p2`, etc.
    *   `m[i][j]` is a 2D array (table) where `m[i][j]` stores the minimum number of scalar multiplications needed to multiply the chain of matrices from `A[i]` to `A[j]` (0-indexed). It's initialized to `0` for single matrices and `infinity` for others.
    *   `s[i][j]` is another 2D array that stores the index `k` where the optimal split occurs for the matrix chain `A[i]...A[j]`. This is used to reconstruct the optimal parenthesization.
    *   The outer loop `for l in range(2, n + 1)` iterates over the chain length `l`. We start with a chain length of 2.
    *   The next nested loop `for i in range(n - l + 1)` iterates over the starting index `i` of the chain.
    *   `j = i + l - 1` calculates the ending index `j` for the current chain length `l`.
    *   The innermost loop `for k in range(i, j)` tries all possible split points `k` between `i` and `j`.
    *   `q = m[i][k] + m[k + 1][j] + p[i] * p[k + 1] * p[j + 1]` calculates the cost of multiplying `A[i]...A[j]` by splitting it at `k`. This cost is the sum of:
        *   The cost to multiply `A[i]...A[k]` (`m[i][k]`).
        *   The cost to multiply `A[k+1]...A[j]` (`m[k+1][j]`).
        *   The cost to multiply the two resulting matrices: `(p[i] x p[k+1]) * (p[k+1] x p[j+1])` which is `p[i] * p[k+1] * p[j+1]`.
    *   If `q` is less than the current `m[i][j]`, it means we found a better way to multiply the subchain, so we update `m[i][j]` and `s[i][j]`.
    *   Finally, `m[0][n-1]` contains the minimum scalar multiplications for the entire chain.

2.  **`print_optimal_parenthesization(s, i, j)` function:**
    *   This is a recursive function that uses the `s` table to reconstruct and print the optimal parenthesization.
    *   The base case is when `i == j`, meaning it's a single matrix, so it returns `A[i+1]` (using 1-based indexing for output clarity).
    *   Otherwise, it retrieves the split point `k` from `s[i][j]` and recursively calls itself for the left and right subproblems, enclosing them in parentheses.
